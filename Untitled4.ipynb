```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
efficiency = df.groupby("Season")[
    "Water_Efficiency_t_per_1000m3"
].mean()

print("Most Water-Efficient Season:", efficiency.idxmax())
print("Water Efficiency:", efficiency.max())

print("Least Water-Efficient Season:", efficiency.idxmin())
print("Water Efficiency:", efficiency.min())
```

```python
for column in [
    "Yield_Tonnes_Ha",
    "Production_Tonnes",
    "Profit_INR",
    "Water_Used_m3",
    "Disease_Pest_Risk_pct"
]:

    print("\n", column)

    q1 = df[column].quantile(0.25)
    q3 = df[column].quantile(0.75)
    iqr = q3 - q1

    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr

    unusual = df[
        (df[column] < lower) |
        (df[column] > upper)
    ]

    print("Number of unusual observations:",
          len(unusual))
```

```python
season_score = df.groupby("Season").agg({
    "Yield_Tonnes_Ha": "mean",
    "Water_Efficiency_t_per_1000m3": "mean",
    "Profit_INR": "mean",
    "Disease_Pest_Risk_pct": "mean"
})

# Normalize values
for col in season_score.columns:
    season_score[col] = (
        season_score[col] - season_score[col].min()
    ) / (
        season_score[col].max() - season_score[col].min()
    )

# Higher yield, efficiency and profit are better
# Lower disease risk is better
season_score["Overall_Score"] = (
    season_score["Yield_Tonnes_Ha"] +
    season_score["Water_Efficiency_t_per_1000m3"] +
    season_score["Profit_INR"] +
    (1 - season_score["Disease_Pest_Risk_pct"])
) / 4

print("Overall Seasonal Performance Score:")
print(season_score.sort_values(
    "Overall_Score",
    ascending=False
))
```
